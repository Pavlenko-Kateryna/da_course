7. Прийміть рішення, чи відхиляєте ви гіпотезу $H_0$. Для прийняття рішення зробіть і порівняння z-статистики з критичним значенням, і проаналізуйте p-value.

```python
sample_mean = np.mean(data)
population_mean = 20
population_std = np.std(data, ddof=1)
sample_size = data.size
alpha = 0.05
```

H0: μ = 20

Ha: μ > 20

Правосторонній тест

**Ручний розрахунок у Python**


```python
SE = population_std / np.sqrt(sample_size)
SE
```

```python
z_stat = (sample_mean - population_mean)/SE
z_stat
```

```python
z_cr = stats.norm.ppf(1- alpha) # правосторонній тест
z_cr
```

Z розраховане менше за критичне, тому не відхиляємо нульову гіпотезу

**через stats та p-value**

```python
p_value = 1 - stats.norm.cdf(z_stat)
p_value
```

```python
if p_value > alpha:
  print("Не відхиляємо нульову гіпотезу")
elif p_value < alpha:
  print("Відхиляємо нульову гіпотезу")
```

**Відповідь**:
Не відхиляємо нульову гіпотезу. Середній чек не підвищено.
