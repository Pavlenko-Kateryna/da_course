2. Як змінюється популярність товарів у різні дні тижня?

```python
df["Date"] = pd.to_datetime(df["Date"])
df["Day"] = df.Date.dt.day_name()
result = df.pivot_table(index = 'Day', columns = 'Product line', values = 'Quantity', aggfunc='sum')

days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

pivot = result.reindex(days)
pivot
```

Кожна категорія має різну популярність протягом тижня. Можна у найбільш непопулярний день робити акції на певний товар з непопулярної категорії, піднімаючи продажі на дану категорію
