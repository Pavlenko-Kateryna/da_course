При цьому функція перед діленням перевіряє, чи другий аргумент не дорівнює 0.   3. Дайте відповідь на питання: "Чому в деяких випадках функція повернула None?"

У деяких випадках функція повернула None, тому що у ній є умова return, але тільки якщо умови виконані (тип int і b != 0), але є випадки у яких якась з умов не виконується і так як в нас є return, але для цих випадків він не прописаний, а функція нічого не повертає, то повертається none (а також друкується print, який у нас прописаний)

````python
def int_division(a: int, b: int):
  if type(a) != int or type(b) != int:
    print("Не працюю з такими типами даних")
  else:
    if b != 0:
      return a / b
    else:
      print("Не можу виконати ділення на 0")
print(int_division(12,6))
print(int_division(20,0))
print(int_division(112,10))
print(int_division(99,0))
print(int_division(11,1.1))
````
